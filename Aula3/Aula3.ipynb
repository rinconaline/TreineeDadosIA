```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("hantavirus_detection_dataset_preprocessed2.csv")
df
```

```python
df = df.drop(columns=['Patient_ID'])
df = df.drop(columns=['Immunological_Risk'])
df = df.drop(columns=['Disease_Onset_Date'])
df = df.drop(columns=['Sample_Date'])
df = df.drop(columns=['Gender'])
df = df.drop(columns=['Age'])
df = df.drop(columns=['Hantavirus_Type'])
df
```

```python
df.info()
```

```python
df.memory_usage(deep=True)
```

```python
df['Symptoms'].unique()
```

```python
df['Symptoms'] = df['Symptoms'].map({'Fever': 1, 'Headache': 2, 'Muscle_Pain': 3, 'Nausea': 4, 'Vomiting': 5, 'Diarrhea': 6, 'Abdominal_Pain': 7, 'Cough': 8, 'Shortness_of_Breath': 9, 'Chills': 10, 'Dizziness': 11, 'Rash': 12, 'Bleeding': 13})
```

```python
df['Renal_Risk'] = df['Renal_Risk'].map({'Low': 1, 'Medium': 2, 'High': 3})
```

```python
df['Disease_Severity'] = df['Disease_Severity'].map({'Mild': 1, 'Moderate': 2, 'Severe': 3})
```

```python
df['Region'] = df['Region'].map({'North': 1, 'South': 2, 'East': 3, 'West': 4, 'Central': 5})
```

```python
df['Exposure_Type'] = df['Exposure_Type'].map({'Occupational': 1, 'Recreational': 2, 'Residential': 3, 'Unknown': 4})
```

```python
import matplotlib.pyplot as plt

casos_positivos = df[df['Hantavirus_Positive'] == 1]['Region'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(casos_positivos.index, casos_positivos.values, color='crimson', edgecolor='black')

plt.title('Distribuição de Casos Positivos de Hantavírus por Região', fontsize=14, fontweight='bold')
plt.xlabel('Região', fontsize=12)
plt.ylabel('Quantidade de Casos Positivos', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

exposicoes = df['Exposure_Type'].dropna().unique()
dados_boxplot = [df[df['Exposure_Type'] == exp]['Liver_Injury_Score'].dropna() for exp in exposicoes]

plt.boxplot(dados_boxplot, labels=exposicoes, patch_artist=True, 
            boxprops=dict(facecolor='lightblue', color='blue'),
            medianprops=dict(color='red', linewidth=2))

plt.title('Score de Lesão Hepática por Tipo de Exposição', fontsize=14, fontweight='bold')
plt.xlabel('Tipo de Exposição', fontsize=12)
plt.ylabel('Liver Injury Score', fontsize=12)
plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
plt.figure(figsize=(10, 6))

cores = {'Severe': 'red', 'Moderate': 'orange', 'Low': 'green'}

df_filtrado = df.dropna(subset=['Disease_Severity'])

for gravidade, grupo in df_filtrado.groupby('Disease_Severity'):
    plt.scatter(
        grupo['WBC_Count_K/uL'], 
        grupo['Platelet_Count_K/uL'], 
        label=gravidade, 
        color=cores.get(gravidade, 'blue'), 
        s=80, 
        edgecolor='black', 
        alpha=0.8
    )

plt.title('Relação WBC vs Plaquetas por Gravidade da Doença', fontsize=14, fontweight='bold')
plt.xlabel('Contagem de Glóbulos Brancos (WBC_Count_K/uL)', fontsize=12)
plt.ylabel('Contagem de Plaquetas (Platelet_Count_K/uL)', fontsize=12)
plt.legend(title='Gravidade')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
```

```python

dados_rins = df['Renal_Risk'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    dados_rins.values, 
    labels=dados_rins.index, 
    autopct='%1.1f%%', 
    startangle=90, 
    colors=['lightskyblue', 'salmon'],
    wedgeprops=dict(edgecolor='black', linewidth=1) 
)

plt.title('Proporção de Pacientes por Nível de Risco Renal', fontsize=14, fontweight='bold')
plt.show()
```
